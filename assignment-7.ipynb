```python
import math
import matplotlib.pyplot as plt

# Minimax Algorithm
def minimax(node, maximizing_player):

    # Leaf node
    if not isinstance(node, list):
        return node

    if maximizing_player:
        best = -math.inf

        for child in node:
            value = minimax(child, False)
            best = max(best, value)

        return best

    else:
        best = math.inf

        for child in node:
            value = minimax(child, True)
            best = min(best, value)

        return best

# Number Tree
tree = [
    [
        [3, 5],
        [2, 9]
    ],
    [
        [12, 5],
        [23, 4]
    ]
]

# Calculate Minimax result
result = minimax(tree, True)

print("Number Tree:")
print(tree)
print("\nMinimax Result:", result)

# Draw Number Tree
fig, ax = plt.subplots(figsize=(13, 7))

# Positions of nodes
pos = {
    "root": (6, 5),

    "n1": (3, 3.8),
    "n2": (9, 3.8),

    "n3": (1.5, 2.5),
    "n4": (4.5, 2.5),
    "n5": (7.5, 2.5),
    "n6": (10.5, 2.5),

    "a": (1, 1),
    "b": (2, 1),
    "c": (4, 1),
    "d": (5, 1),

    "e": (7, 1),
    "f": (8, 1),
    "g": (10, 1),
    "h": (11, 1)
}

# Values inside nodes
values = {
    "root": "12",
    "n1": "5",
    "n2": "12",

    "n3": "5",
    "n4": "9",
    "n5": "12",
    "n6": "23",

    "a": "3",
    "b": "5",
    "c": "2",
    "d": "9",

    "e": "12",
    "f": "5",
    "g": "23",
    "h": "4"
}

# Connections
edges = [
    ("root", "n1"),
    ("root", "n2"),

    ("n1", "n3"),
    ("n1", "n4"),

    ("n2", "n5"),
    ("n2", "n6"),

    ("n3", "a"),
    ("n3", "b"),

    ("n4", "c"),
    ("n4", "d"),

    ("n5", "e"),
    ("n5", "f"),

    ("n6", "g"),
    ("n6", "h")
]

# Draw connections
for start, end in edges:
    x1, y1 = pos[start]
    x2, y2 = pos[end]

    ax.plot(
        [x1, x2],
        [y1, y2],
        linewidth=1.5
    )

# Draw number nodes
for node, (x, y) in pos.items():

    ax.scatter(
        x, y,
        s=1000,
        facecolors="white",
        edgecolors="black",
        linewidths=2,
        zorder=2
    )

    ax.text(
        x, y,
        values[node],
        ha="center",
        va="center",
        fontsize=12,
        fontweight="bold",
        zorder=3
    )


# MAX / MIN outside the tree
ax.text(
    0.2, 5,
    "MAX",
    fontsize=14,
    fontweight="bold",
    ha="center"
)

ax.text(
    0.2, 3.8,
    "MIN",
    fontsize=14,
    fontweight="bold",
    ha="center"
)

ax.text(
    0.2, 2.5,
    "MAX",
    fontsize=14,
    fontweight="bold",
    ha="center"
)

ax.text(
    0.2, 1,
    "LEAF",
    fontsize=14,
    fontweight="bold",
    ha="center"
)


# Title
ax.set_title(
    "Adversarial Search using Minimax",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlim(-1, 12)
ax.set_ylim(0, 5.7)

ax.axis("off")

plt.show()
```
